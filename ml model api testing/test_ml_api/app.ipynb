```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.naive_bayes import GaussianNB

from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score



```

```python
df = sns.load_dataset("tips")
df.drop(columns=["day"], inplace=True)
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
cat_col = df.select_dtypes(include="category")
num_col = df.select_dtypes(include=["int64", "float64"])
```

```python
for i in cat_col:
    le = LabelEncoder()
    df[i] = le.fit_transform(df[i])
    joblib.dump(le, f"le_{i}.pkl")
    
```

```python
x = df.drop(columns=["smoker"])
y = df["smoker"]
```

```python
x
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
models = {

    "Logistic Regression": LogisticRegression(),

    "Decision Tree": DecisionTreeClassifier(max_depth=20),

    "Random Forest": RandomForestClassifier(n_estimators=50, max_depth=20, max_features="sqrt"),

    "Naive Bayes": GaussianNB(),

    "SVC" : SVC(kernel="linear", C=1.0),

    "KNN" : KNeighborsClassifier(n_neighbors=5)

}
```

```python
results = {}
```

```python
for name, i in models.items():
    i.fit(x_train, y_train)
    y_pred = i.predict(x_test)
    results[name] = accuracy_score(y_test, y_pred)
    print(name, accuracy_score(y_test, y_pred))
```

```python
best_model_name = max(results, key=results.get)
best_model = models[best_model_name]
```

```python
joblib.dump(best_model, "best_model.pkl")
```
